```
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('PySpark Practice').getOrCreate()
```

```
# Sample data
data = [
("USA", "North America", 100, 50.5),
("India", "Asia", 300, 20.0),
("Germany", "Europe", 200, 30.5),
("Australia", "Oceania", 150, 60.0),
("Japan", "Asia", 120, 45.0),
("Brazil", "South America", 180, 25.0)
]

# Define the schema
columns = ["Country", "Region", "UnitsSold", "UnitPrice"]
```

```
# Create DataFrame

df = spark.createDataFrame(data= data, schema = columns)
df.show()
```

```
df.printSchema()
```

### Sorting a single Column

```
from pyspark.sql.functions import *
df.orderBy(desc('UnitsSold')).show()
```

### Sorting by mutiple columns

```
df.orderBy('Country', 'UnitsSold').show()
```

### Sorting by desc and limit

```
df.orderBy(desc('UnitsSold')).limit(3).show()
```

## String Functions

```
#Convert the first letter of each word to uppercase
df.select(initcap(col('Region')).alias('Region')).show()
```

```
#Convert all text to lowercase

df.select(lower(col('Country')).alias('Country')).show()
```

```
# Convert all text to uppercase
df.select(upper(col('Country')).alias('Country')).show()
```

```
# Concatenate two columns without separator
df.select(concat(col('Country'), col('Region')).alias('Concated_Location')).show()
```

```
# Concatenate two columns with separator
df.select(concat_ws(' | ', col('Region'), col('Country')).alias('Concated_Location')).show()
```

## Date Functions


```
df = spark.range(10)
```

```
df = df.withColumn('Today', current_date())\
  .withColumn('now', current_timestamp())
```

```
emp_data = [
Row(emp_id=1, emp_name="Alice", emp_salary=50000, 
emp_dept_id=101, emp_location="New York"),
Row(emp_id=2, emp_name="Bob", emp_salary=60000, 
emp_dept_id=102, emp_location="Los Angeles"),
Row(emp_id=3, emp_name="Charlie", emp_salary=55000, 
emp_dept_id=101, emp_location="Chicago"),
Row(emp_id=4, emp_name="David", emp_salary=70000, 
emp_dept_id=103, emp_location="San Francisco"),
Row(emp_id=5, emp_name="Eve", emp_salary=48000, 
emp_dept_id=102, emp_location="Houston")
]

dept_data = [
Row(dept_id=101, dept_name="Engineering", dept_head="John", 
dept_location="New York"),
Row(dept_id=102, dept_name="Marketing", dept_head="Mary", 
dept_location="Los Angeles"),
Row(dept_id=103, dept_name="Finance", dept_head="Frank", 
dept_location="Chicago")
]

emp_columns = ["emp_id", "emp_name", "emp_salary", "emp_dept_id", 
"emp_location"]
dept_columns = ["dept_id", "dept_name", "dept_head", 
"dept_location"]
```

```
schema = StructType({
    StructField('Name', StringType(), True),
    StructField('Age', IntegerType(), True),
    StructField('Salary', IntegerType(), True)
})

data = [
("Alice", 25, 3000),
("Bob", 35, 4000),
("Charlie", 40, 5000),
("David", 28, 4500),
("Eve", 32, 3500)
]

df = spark.createDataFrame(data=data, schema=schema)
```

```
df.show(truncate=False)
```

```
df = df.withColumnRenamed('Agee', 'Age')
```

```
df = df.withColumn('Status', when(col('Age') < 30, 'Young').otherwise('Adult'))\
    .withColumn('Income Bracket', when(col('Salary') < 4000, 'Low').when((col('Salary') >= 4000) & (col('Salary') <= 4500), 'Medium').otherwise('High'))
```

```
df.show(truncate=False)
```

## Window Functions

```
data = [
("A", "X", 1, "2023-01-01"),
("A", "X", 2, "2023-01-02"),
("A", "Y", 3, "2023-01-01"),
("A", "Y", 3, "2023-01-02"),
("B", "X", 5, "2023-01-01"),
("B", "X", 4, "2023-01-02"),
]

columns = ["category", "sub_category", "value", "timestamp"]

df = spark.createDataFrame(data = data, schema = columns)
```

```
df.show(truncate=False)
```

```
window_spec = Window.partitionBy('category', 'sub_category').orderBy('timestamp', 'value')
```

```
df = df.withColumn('row_number', row_number().over(window_spec))
df = df.withColumn('rank', rank().over(window_spec))
df = df.withColumn('dense_rank', dense_rank().over(window_spec))
df = df.withColumn('rank', rank().over(window_spec))
```

```
data = [
("Alice", 100),
("Bob", 200),
("Charlie", 200),
("David", 300),
("Eve", 400),
("Frank", 500),
("Grace", 500),
("Hank", 600),
("Ivy", 700),
("Jack", 800)
]

columns = ["Name", "Score"]

df = spark.createDataFrame(data=data, schema=columns)
```

```

df.withColumn('rank', rank().over(Window.orderBy('Score'))).show()
```

```
window_spec = Window.orderBy('Score')

df = df.withColumn('rank', rank().over(window_spec))
```

```
df = df.withColumn('row_number', row_number().over(Window.orderBy('Score')))
df = df.withColumn('dense_rank', dense_rank().over(Window.orderBy('Score')))
```

```
df.show(truncate=False)
```

```
df.withColumn('ScoreDifference', lead('Score').over(Window.orderBy('Score')) - df['Score']).show()
```
